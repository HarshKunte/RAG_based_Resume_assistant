# Resume RAG Assistant Demo

Enter a job description, retrieve matching resumes, and show performance metrics.

```python
from pathlib import Path
import json
import sys

ROOT = Path.cwd()
if not (ROOT / 'job_matcher.py').exists():
    raise FileNotFoundError('Run this notebook from the repository root.')
sys.path.insert(0, str(ROOT))
from job_matcher import get_top_matches
from evaluation import evaluate
```

## 1. Enter A Job Description

```python
# Change this query to experiment.
job_description = (
    'Required candidate should have 4+ years of professional experience '
    'in React and Python, with frontend performance experience.'
)
print(job_description)
```

## 2. Get Matching Resumes

```python
result = get_top_matches(job_description, top_k=5)

for rank, match in enumerate(result['top_matches'], start=1):
    print(f'{rank}. {match["candidate_name"]}')
    print(f'   Score: {match["match_score"]}/100')
    print(f'   Matched skills: {", ".join(match["matched_skills"]) or "None"}')
    print(f'   Resume: {match["resume_path"]}')
    print(f'   Reasoning: {match["reasoning"]}\n')

# Uncomment this line to show the exact required JSON response.
# print(json.dumps(result, indent=2))
```

## Summary
The system combines semantic search and BM25 keyword search, groups results by candidate, applies the experience requirement, calculates the score, and uses an LLM to explain the matching evidence.
